```python
import warnings
# Скачаем датасет Gazeta

import os
from datasets import load_dataset
import json
os.environ["HF_HUB_DISABLE_SYMLINKS_WARNING"] = "1"
warnings.filterwarnings("ignore", message=".*symlinks.*")

os.makedirs("gazeta", exist_ok=True)

# Скачиваем validation и test

for split in ["test", "validation"]:
    dataset = load_dataset("IlyaGusev/gazeta", split=split)

    texts = [item["text"] for item in dataset]
    refs = [item["summary"] for item in dataset]

    with open(f"gazeta/{split}_texts.json", "w", encoding="utf-8") as f:
        json.dump(texts, f, ensure_ascii=False)
    with open(f"gazeta/{split}_refs.json", "w", encoding="utf-8") as f:
        json.dump(refs, f, ensure_ascii=False)

    print(f"{split}: {len(texts)} примеров")
```
